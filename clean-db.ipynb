```python
import pandas as pd
import glob
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
import ast
import re

# Find all CSV files that start with "Node"
files = glob.glob("Node*.csv")
node_data = {}

# Load all files into memory with a progress bar
for file in tqdm(files, desc="Loading CSVs"):
    # Extracts only the digits that come directly after 'Node'
    match = re.search(r'Node(\d+)', file)
    
    if match:
        node_number = match.group(1) # This stores "1", "2", "10", etc.
        node_data[node_number] = pd.read_csv(file)
```

```python
import pandas as pd
import glob
from tqdm import tqdm
import re

# 1. Define the exact columns you want to remove
columns_to_drop = [
    'lat', 
    'lng', 
    'excursion_peak_to_peak_hz', 
    'excursion_peak_deviation_hz', 
    'excursion_rms_deviation_hz', 
    'depth_peak_to_peak', 
    'depth_peak_deviation', 
    'depth_rms_deviation'
]

# 2. Find the original files again
files = glob.glob("Node*.csv")

# 3. Loop through, drop columns, and overwrite the files
for file in tqdm(files, desc="Cleaning and Overwriting CSVs"):
    # Load the current file
    df = pd.read_csv(file)
    
    # Drop the specified columns (errors='ignore' prevents crashes if a column is missing)
    df = df.drop(columns=columns_to_drop, errors='ignore')
    
    # Save the cleaned dataframe back to the EXACT SAME filename (overwriting it)
    # index=False prevents pandas from adding an extra unnamed column of row numbers
    df.to_csv(file, index=False)
    
    # Update your in-memory dictionary so you can keep working without reloading
    match = re.search(r'Node(\d+)', file)
    if match:
        node_number = match.group(1)
        node_data[node_number] = df

print("\nAll datasets have been cleaned and the original CSV files are successfully updated!")
```
