```python
import requests
import json

campaign_id = 176
ANE1_mac = "d8:3a:dd:f7:1d:f2"
API_URL = "https://rsm.ane.gov.co:12443/api"
```

# Request acquisition parameters for a specific campaign

```python
url_params = f"{API_URL}/campaigns/{campaign_id}/parameters"

response = requests.get(url_params)
pretty_json = json.dumps(response.json(), indent=4)
print(pretty_json)
```

# Request realtime signals for a sensor

```python
url_realtime = f"{API_URL}/campaigns/sensor/{ANE1_mac}/realtime"
response = requests.get(url_realtime)
pretty_json = json.dumps(response.json(), indent=4)
print(pretty_json)
```

# Request signals from a sensor in a certain campaign ID

```python
def filter_signals_from_camp_and_mac(base_url, mac, camp_id):
    url = f"{base_url}/campaigns/sensor/{mac}/signals"
    params = {"campaign_id": camp_id, "page": 1, "page_size": 5000}
    all_signals = []
    
    while True:
        data = requests.get(url, params=params).json()
        all_signals.extend(data['measurements'])
        
        if not data['pagination']['has_next']:
            break
            
        params['page'] += 1

    return all_signals

# How to use:
signals = filter_signals_from_camp_and_mac(API_URL, ANE1_mac, campaign_id)

print(f"Total signals retrieved: {len(signals)}")

print("First signal:")
print(signals[0])
```
